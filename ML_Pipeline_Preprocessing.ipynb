```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('datasets/watchbase_data_raw_scrapy.csv')
df.head()
```

```python
df.info()
df.isna().sum()
```

```python
df.columns = df.columns.str.replace('_', ' ')
```

```python
fig, ax = plt.subplots(figsize=(10, 10))
sns.heatmap(df.isna(), cbar=False, cmap='Blues', yticklabels=False)
plt.title('Missing Data Heatmap', fontsize=16)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
df = df.drop(columns=['Produced'])
```

```python
df['Case Material'].value_counts()
```

```python
df['Case Material'] = df['Case Material'].apply(lambda x: 'Stainless Steel' if pd.isna(x) else x)
df['Case Material'].value_counts()
```

```python
df = df.drop(columns=['Lug Width'])
```

```python
df['Water Resistance'].value_counts()
```

```python
df['Water Resistance'] = df['Water Resistance'].str.replace(' m', '').astype(float)
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['Water Resistance'].dropna(), kde=True)
```

```python
# value_counts = df['Water Resistance'].dropna().value_counts(normalize=True)

# df.loc[df['Water Resistance'].isna(), 'Water Resistance'] = np.random.choice(
#     value_counts.index,
#     size=df['Water Resistance'].isna().sum(),
#     p=value_counts.values
# )
```

```python
df['Water Resistance'] = df['Water Resistance'].fillna(df['Water Resistance'].mode()[0])
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df['Water Resistance'].dropna(), kde=True)
```

```python
df = df.drop(columns=['Dial Finish'])
```

```python
df['Dial Indexes'].value_counts()
```

```python
df['Dial Hands'] = df['Dial Hands'].fillna(df['Dial Hands'].mode()[0])
```

```python
df['Price'] = df['Price'].astype(float)
```
